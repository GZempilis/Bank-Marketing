```python
%load_ext autoreload
%autoreload 2
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python


df = pd.read_csv('../data/bank-additional-full.csv', sep=';')


df_raw = df.drop(columns=['duration'])


sns.set_theme(style="whitegrid")


numerical_cols = df_raw.select_dtypes(include=['int64', 'float64']).columns


df_raw[numerical_cols].hist(figsize=(15, 10), bins=30, edgecolor='black', color='skyblue')
plt.suptitle('Histograms of Numerical Variables (Raw Data)', fontsize=16)
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()


categorical_cols = df_raw.select_dtypes(include=['object']).columns


n_cat = len(categorical_cols)
fig, axes = plt.subplots(nrows=(n_cat + 1) // 2, ncols=2, figsize=(16, 22))
axes = axes.flatten()

for i, col in enumerate(categorical_cols):
    
    sns.countplot(
        data=df_raw, 
        y=col, 
        ax=axes[i], 
        order=df_raw[col].value_counts().index, 
        palette='viridis'
    )
    axes[i].set_title(f'Distribution of {col}', fontsize=13)
    axes[i].set_xlabel('Count')
    axes[i].set_ylabel('')


for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```
